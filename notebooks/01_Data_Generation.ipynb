```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta

fake = Faker("en_IN")

blood_groups = ["A+", "A-", "B+", "B-", "AB+", "AB-", "O+", "O-"]

occupations = [
    "Student", "Engineer", "Teacher", "Doctor", "Farmer",
    "Business", "Software Developer", "Nurse", "Driver",
    "Lawyer", "Accountant", "Police", "Chef", "Sales Executive"
]

cities = {
    "Hyderabad": "Telangana",
    "Vijayawada": "Andhra Pradesh",
    "Visakhapatnam": "Andhra Pradesh",
    "Bengaluru": "Karnataka",
    "Chennai": "Tamil Nadu",
    "Mumbai": "Maharashtra",
    "Delhi": "Delhi",
    "Kolkata": "West Bengal",
    "Pune": "Maharashtra",
    "Jaipur": "Rajasthan"
}

num_records = 100000

patients = []

for i in range(1, num_records + 1):

    city = random.choice(list(cities.keys()))

    height = random.randint(145, 190)
    weight = random.randint(40, 110)

    bmi = round(weight / ((height / 100) ** 2), 1)

    registration = fake.date_between(start_date="-5y", end_date="today")

    patients.append({
        "Patient_ID": f"P{i:06}",
        "Full_Name": fake.name(),
        "Age": random.randint(1, 90),
        "Gender": random.choice(["Male", "Female", "Other"]),
        "Blood_Group": random.choice(blood_groups),
        "Height_cm": height,
        "Weight_kg": weight,
        "BMI": bmi,
        "Occupation": random.choice(occupations),
        "City": city,
        "State": cities[city],
        "Phone": fake.phone_number(),
        "Email": fake.email(),
        "Insurance": random.choice(["Yes", "No"]),
        "Registration_Date": registration
    })

patients_df = pd.DataFrame(patients)

patients_df.head()
```

```python
patients_df.to_csv("../data/raw/patients.csv", index=False)

print("patients.csv created successfully!")
print(patients_df.shape)
```
