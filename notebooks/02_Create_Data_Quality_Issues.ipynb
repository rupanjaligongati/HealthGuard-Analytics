```python
import pandas as pd
import numpy as np
import random

print("Libraries imported successfully!")
```

```python
patients_df = pd.read_csv(
    r"C:\Users\ajayg\Documents\GitHub\HealthGuard-Analytics\data\raw\patients.csv"
)

print("Dataset Loaded Successfully!")
print("Shape:", patients_df.shape)

patients_df.head()
```

```python
columns = ["Phone", "Email", "BMI"]

for col in columns:
    patients_df.loc[
        patients_df.sample(frac=0.03, random_state=42).index,
        col
    ] = np.nan

print("Missing values added successfully!")
```

```python
duplicates = patients_df.sample(
    n=500,
    random_state=42
)

patients_df = pd.concat(
    [patients_df, duplicates],
    ignore_index=True
)

print("Duplicate records added!")
print("New Shape:", patients_df.shape)
```

```python
invalid_index = patients_df.sample(
    n=100,
    random_state=42
).index

patients_df.loc[invalid_index, "Age"] = np.random.choice(
    [-5, 150],
    size=100
)

print("Invalid ages created!")
```

```python
patients_df["City"] = patients_df["City"].replace({
    "Hyderabad": "Hyderbad",
    "Bengaluru": "Banglore",
    "Visakhapatnam": "Vizag"
})

print("City names modified!")
```

```python
outlier_index = patients_df.sample(
    n=50,
    random_state=42
).index

patients_df.loc[outlier_index, "BMI"] = 75

print("BMI outliers added!")
```

```python
print("Dataset Shape:", patients_df.shape)

print("\nMissing Values:")
print(patients_df.isnull().sum())

print("\nDuplicate Records:")
print(patients_df.duplicated().sum())

print("\nInvalid Ages:")
print(patients_df[(patients_df["Age"] < 0) | (patients_df["Age"] > 120)].shape[0])
```

```python
patients_df.to_csv(
    r"C:\Users\ajayg\Documents\GitHub\HealthGuard-Analytics\data\raw\patients_dirty.csv",
    index=False
)
```
