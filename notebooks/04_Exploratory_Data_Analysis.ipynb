## Analysis 3: Blood Group Distribution

```python
blood = patients_df["Blood_Group"].value_counts()

print(blood)

blood_percentage = round((blood / blood.sum()) * 100, 2)

print(blood_percentage)

plt.figure(figsize=(8,5))

ax = blood.plot(kind="bar")

plt.title("Patient Blood Group Distribution", fontsize=14)

plt.xlabel("Blood Group", fontsize=12)

plt.ylabel("Number of Patients", fontsize=12)

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.5)

# Display values above bars
for i, value in enumerate(blood):
    plt.text(i, value + 80, str(value), ha="center", fontsize=9)

plt.show()
```

## Business Insight

This analysis identifies the distribution of blood groups among registered patients. Understanding blood group availability helps hospitals and blood banks plan inventory and respond effectively during emergencies.

## Key Findings

- AB+ is the most common blood group in this dataset (12,619 patients).
- AB- is the least common blood group (12,380 patients).
- The distribution of blood groups is fairly balanced across all categories.
- Hospitals and blood banks should maintain stock for every blood group, while giving slightly higher priority to more common blood groups.

## Recommendation

Hospitals should monitor blood inventory regularly and ensure adequate availability of all blood groups. Emergency preparedness plans should include both common and rare blood types.

# Analysis 8: BMI Category Distribution

## Objective

To classify patients into standard BMI categories and analyze the distribution of body weight status. This analysis helps healthcare providers identify high-risk groups and supports preventive healthcare planning, nutrition counseling, and lifestyle intervention programs.

```python
# Create BMI Categories

patients_df["BMI_Category"] = pd.cut(
    patients_df["BMI"],
    bins=[0, 18.5, 24.9, 29.9, 100],
    labels=["Underweight", "Normal Weight", "Overweight", "Obese"]
)

# Count Categories
bmi_category = patients_df["BMI_Category"].value_counts()

print(bmi_category)

# Percentage
bmi_percentage = round((bmi_category / bmi_category.sum()) * 100, 2)

print("\nPercentage Distribution:")
print(bmi_percentage)

# Plot
plt.figure(figsize=(9,6))

ax = bmi_category.plot(kind="bar")

plt.title("BMI Category Distribution", fontsize=14)

plt.xlabel("BMI Category", fontsize=12)

plt.ylabel("Number of Patients", fontsize=12)

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.5)

# Display values
for i, value in enumerate(bmi_category):
    plt.text(i, value + 200, str(value), ha="center", fontsize=10)

plt.show()
```

## Business Insight

BMI category analysis provides a comprehensive view of the overall health profile of the patient population. Identifying the distribution of underweight, normal weight, overweight, and obese patients enables healthcare providers to prioritize preventive care, promote healthy lifestyle interventions, and reduce the long-term burden of chronic diseases.

## Key Findings

- **Obese** is the largest BMI category, accounting for **35,390 patients (35.39%)**, indicating a significant proportion of the patient population is at increased risk of obesity-related health conditions.
- **24,767 patients (24.77%)** fall within the **Normal Weight** category, while **22,163 patients (22.16%)** are **Overweight** and **17,680 patients (17.68%)** are **Underweight**.
- More than **57% of patients** are classified as **Overweight or Obese**, highlighting the importance of preventive healthcare initiatives focused on weight management and healthy lifestyles.
- BMI category analysis provides valuable insights for identifying high-risk populations and supports evidence-based planning for nutrition counseling, fitness programs, and chronic disease prevention.

## Recommendation

Healthcare organizations should strengthen preventive healthcare initiatives by promoting balanced nutrition, regular physical activity, and routine BMI assessments. Priority should be given to overweight and obese individuals through personalized lifestyle counseling and early intervention programs to reduce the risk of diabetes, hypertension, cardiovascular diseases, and other obesity-related conditions.

## Healthcare Impact

BMI category analysis enables healthcare providers to identify high-risk patient groups, allocate preventive healthcare resources effectively, and design targeted wellness programs. These insights support early intervention strategies, improve population health outcomes, and contribute to reducing long-term healthcare costs associated with chronic diseases.

# Analysis 9: Age Group Distribution

## Objective

To classify patients into different age groups and analyze their distribution. This analysis helps healthcare providers understand which age groups utilize healthcare services the most and supports age-specific healthcare planning, preventive care programs, and resource allocation.

```python
# Create Age Groups

patients_df["Age_Group"] = pd.cut(
    patients_df["Age"],
    bins=[0, 17, 35, 60, 100],
    labels=[
        "Children (0-17)",
        "Young Adults (18-35)",
        "Adults (36-60)",
        "Senior Citizens (60+)"
    ]
)

# Count Age Groups
age_group = patients_df["Age_Group"].value_counts()

print(age_group)

# Percentage
age_percentage = round((age_group / age_group.sum()) * 100, 2)

print("\nPercentage Distribution:")
print(age_percentage)

# Plot
plt.figure(figsize=(10,6))

ax = age_group.plot(kind="bar")

plt.title("Patient Age Group Distribution", fontsize=14)

plt.xlabel("Age Group", fontsize=12)

plt.ylabel("Number of Patients", fontsize=12)

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.5)

# Display values above bars
for i, value in enumerate(age_group):
    plt.text(i, value + 200, str(value), ha="center", fontsize=10)

plt.show()
```

## Business Insight

Age group analysis provides valuable insights into the healthcare needs of different population segments. Understanding the distribution of patients across age groups helps hospitals optimize medical services, allocate healthcare resources efficiently, and develop targeted preventive care programs for children, adults, and senior citizens.

## Key Findings

- **Senior Citizens (60+)** represent the largest age group, accounting for **33,298 patients (33.30%)** of the total patient population.
- **Adults (36–60)** form the second-largest group with **28,016 patients (28.02%)**, followed by **Young Adults (18–35)** with **19,875 patients (19.88%)** and **Children (0–17)** with **18,811 patients (18.81%)**.
- More than **61% of patients** belong to the **Adult and Senior Citizen** age groups, indicating a higher demand for healthcare services among middle-aged and elderly individuals.
- Age group analysis enables healthcare organizations to design age-specific preventive care programs and allocate medical resources more effectively.

## Recommendation

Healthcare organizations should prioritize preventive healthcare programs and regular health screenings for adults and senior citizens, as they represent the majority of the patient population. At the same time, dedicated pediatric healthcare services and vaccination programs should continue to support the healthcare needs of children.

## Healthcare Impact

Age group analysis enables healthcare providers to forecast healthcare demand, improve workforce planning, optimize hospital infrastructure, and implement age-specific wellness initiatives. These insights support evidence-based decision-making and contribute to improved patient care across all age groups.

# Analysis 10: BMI Category by Gender

## Objective

To analyze the distribution of BMI categories across different genders. This analysis helps healthcare organizations identify gender-specific health trends, assess obesity prevalence, and develop targeted wellness and preventive healthcare programs.

```python
# BMI Category by Gender

bmi_gender = pd.crosstab(
    patients_df["Gender"],
    patients_df["BMI_Category"]
)

print(bmi_gender)

# Percentage by Gender
bmi_gender_percentage = round(
    bmi_gender.div(bmi_gender.sum(axis=1), axis=0) * 100,
    2
)

print("\nPercentage Distribution:")
print(bmi_gender_percentage)

# Plot
plt.figure(figsize=(10,6))

ax = bmi_gender.plot(kind="bar")

plt.title("BMI Category by Gender", fontsize=14)

plt.xlabel("Gender", fontsize=12)

plt.ylabel("Number of Patients", fontsize=12)

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.legend(title="BMI Category")

plt.show()
```

## Business Insight

Analyzing BMI categories by gender provides valuable insights into population health trends and helps identify gender-specific healthcare challenges. These findings enable healthcare organizations to design targeted wellness programs, improve preventive healthcare strategies, and optimize resource allocation for high-risk patient groups.

## Key Findings

- **Female patients** have the highest proportion of patients in the **Obese** category (**35.88%**), followed by **Other** (**35.40%**) and **Male** (**34.89%**).
- Across all gender groups, the **Obese** category represents the largest proportion of patients, highlighting obesity as a significant healthcare concern.
- The distributions of **Underweight**, **Normal Weight**, and **Overweight** patients are relatively consistent across all genders, indicating similar BMI patterns within the dataset.
- Gender-based BMI analysis helps healthcare organizations identify high-risk populations and supports the development of targeted wellness programs, nutrition counseling, and preventive healthcare initiatives.

# Analysis 12: Correlation Analysis & Heatmap

## Objective

To analyze the relationships between numerical variables in the patient dataset using a correlation matrix. This analysis helps identify patterns, dependencies, and potential relationships between patient attributes, supporting better healthcare insights and data-driven decision-making.

```python
# Select numerical columns

numerical_data = patients_df.select_dtypes(include=["int64", "float64"])

# Correlation Matrix
correlation = numerical_data.corr()

print(correlation)

# Plot Heatmap

plt.figure(figsize=(8,6))

plt.imshow(correlation, cmap="coolwarm", aspect="auto")

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

# Show correlation values inside the heatmap
for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(
            j,
            i,
            f"{correlation.iloc[i,j]:.2f}",
            ha="center",
            va="center",
            fontsize=9,
            color="black"
        )

plt.title("Correlation Heatmap", fontsize=14)

plt.tight_layout()

plt.show()
```

## Business Insight

Correlation analysis helps healthcare organizations understand how patient health indicators are related. Strong relationships, such as between weight and BMI, can support risk assessment, preventive healthcare planning, and predictive analytics, while weak correlations indicate that multiple factors should be considered when evaluating patient health.

## Key Findings

- The strongest **positive correlation** exists between **Weight_kg** and **BMI** (**0.84**), indicating that BMI increases significantly as body weight increases.
- A **moderate negative correlation** is observed between **Height_cm** and **BMI** (**-0.49**), suggesting that taller individuals tend to have lower BMI values when weight is held relatively constant.
- **Age** shows **very weak correlations** with Height, Weight, and BMI, indicating that age has minimal direct linear relationship with these variables in this dataset.
- Correlation analysis provides valuable insights for healthcare organizations by identifying relationships between patient health indicators and supporting evidence-based clinical decision-making.

## Recommendation

Healthcare providers should use correlation analysis alongside clinical expertise to identify meaningful health patterns and support data-driven decision-making. Monitoring key indicators such as weight and BMI can help detect patients at risk of obesity-related conditions and enable timely preventive interventions.

## Healthcare Impact

Understanding relationships between health indicators enables healthcare organizations to improve patient risk assessment, optimize preventive care strategies, and support personalized healthcare planning. Correlation analysis also forms the foundation for predictive analytics and future machine learning applications in healthcare.
