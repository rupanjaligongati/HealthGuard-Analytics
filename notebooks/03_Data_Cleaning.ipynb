```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
patients_df = pd.read_csv(
    r"C:\Users\ajayg\Documents\GitHub\HealthGuard-Analytics\data\raw\patients_dirty.csv"
)

print("Dataset Loaded Successfully!")
print("Shape:", patients_df.shape)

patients_df.head()
```

```python
patients_df.isnull().sum()
```

```python
patients_df["Phone"] = patients_df["Phone"].fillna("Not Available")

patients_df["Email"] = patients_df["Email"].fillna("unknown@email.com")

patients_df["BMI"] = patients_df["BMI"].fillna(
    patients_df["BMI"].median()
)

print("Missing values handled successfully!")
```

```python
print("Before:", patients_df.shape)

patients_df = patients_df.drop_duplicates()

print("After:", patients_df.shape)
```

```python
patients_df["Patient_ID"].duplicated().sum()
```

```python
patients_df.loc[
    (patients_df["Age"] < 0) |
    (patients_df["Age"] > 120),
    "Age"
] = np.nan

patients_df["Age"] = patients_df["Age"].fillna(
    patients_df["Age"].median()
)

print("Invalid ages corrected!")
```

```python
patients_df["City"] = patients_df["City"].replace({
    "Hyderbad": "Hyderabad",
    "Banglore": "Bengaluru",
    "Vizag": "Visakhapatnam"
})

print("City names corrected!")
```

```python
Q1 = patients_df["BMI"].quantile(0.25)
Q3 = patients_df["BMI"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

patients_df["BMI"] = patients_df["BMI"].clip(lower, upper)

print("BMI outliers handled successfully!")
```

```python
print("Remaining Missing Values")

print(patients_df.isnull().sum())

print()

print("Duplicate Records")

print(patients_df.duplicated().sum())

print()

print("Invalid Ages")

print(
    patients_df[
        (patients_df["Age"] < 0) |
        (patients_df["Age"] > 120)
    ].shape[0]
)
```

```python
patients_df.to_csv(
    r"C:\Users\ajayg\Documents\GitHub\HealthGuard-Analytics\data\cleaned\patients_cleaned.csv",
    index=False
)

print("patients_cleaned.csv created successfully!")
```
